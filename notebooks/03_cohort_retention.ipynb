# 03. Cohort Retention Matrix & Longitudinal Decay
Analyzes monthly signup cohorts, tracking retention over 6-month horizons, and evaluates observational correlation between feature usage and 7-day retention.

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.utils.io import load_dataframe

cohort_mart = load_dataframe('data/marts/mart_cohort_retention.parquet')
matrix = cohort_mart.pivot(index='cohort_month', columns='period_month', values='retention_rate') * 100.0
matrix
```

## 1. Monthly Retention Heatmap

```
fig, ax = plt.subplots(figsize=(9, 5))
sns.heatmap(matrix, annot=True, fmt='.1f', cmap='Blues', ax=ax)
ax.set_title('Signup Cohort Monthly Retention (%)')
plt.tight_layout()
plt.show()
```
