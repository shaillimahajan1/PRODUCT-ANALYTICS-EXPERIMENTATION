# 01. FlowPulse Data Exploration & Schema Validation
This notebook inspects the raw event stream, validates data quality invariants, and explores user demographics, device distributions, and session characteristics.

```
import duckdb
import pandas as pd
from src.utils.config import load_config

config = load_config()
con = duckdb.connect(config['paths']['database'])
con.execute('SHOW TABLES').df()
```

## 1. User Demographics & Plan Distribution

```
users_df = con.execute('SELECT country, device_type, acquisition_channel, COUNT(*) as user_count FROM dim_users GROUP BY country, device_type, acquisition_channel ORDER BY user_count DESC LIMIT 10').df()
users_df
```

## 2. Event Vocabulary & Volume

```
con.execute('SELECT event_name, COUNT(*) as event_count, COUNT(DISTINCT user_id) as unique_users FROM fct_events GROUP BY event_name ORDER BY event_count DESC').df()
```
