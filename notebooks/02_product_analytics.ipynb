# 02. Product Funnel & Milestone Conversion Analysis
Evaluates multi-step product conversion from registration to paid purchase, identifying major drop-off bottlenecks and segment variances.

```
import pandas as pd
import matplotlib.pyplot as plt
from src.utils.io import load_dataframe

funnel_df = load_dataframe('data/marts/mart_funnel.parquet')
funnel_df
```

## 1. Funnel Conversion & Drop-off Rates

```
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.barh(funnel_df['stage_name'], funnel_df['users_reached'], color='#2563EB')
ax.set_title('Product Funnel Progression')
ax.set_xlabel('Unique Users')
plt.tight_layout()
plt.show()
```

## 2. Segment-level Funnel by Device

```
device_funnel = load_dataframe('data/marts/mart_funnel_by_device.parquet')
device_funnel.pivot(index='stage_name', columns='device_type', values='stage_conversion_rate')
```
